```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, random_split
import torchvision.transforms.functional as TF
import random
import cv2
import numpy as np
from PIL import Image
import os

# 1. Model Definition
class DoubleConv(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.double_conv = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True)
        )
    def forward(self, x): return self.double_conv(x)

class WaterloggingUNet(nn.Module):
    def __init__(self, n_channels=3, n_classes=1):
        super(WaterloggingUNet, self).__init__()
        self.inc = DoubleConv(n_channels, 64)
        self.down1 = nn.Sequential(nn.MaxPool2d(2), DoubleConv(64, 128))
        self.down2 = nn.Sequential(nn.MaxPool2d(2), DoubleConv(128, 256))
        self.down3 = nn.Sequential(nn.MaxPool2d(2), DoubleConv(256, 512))
        self.down4 = nn.Sequential(nn.MaxPool2d(2), DoubleConv(512, 1024))
        self.up1 = nn.ConvTranspose2d(1024, 512, kernel_size=2, stride=2)
        self.conv1 = DoubleConv(1024, 512)
        self.up2 = nn.ConvTranspose2d(512, 256, kernel_size=2, stride=2)
        self.conv2 = DoubleConv(512, 256)
        self.up3 = nn.ConvTranspose2d(256, 128, kernel_size=2, stride=2)
        self.conv3 = DoubleConv(256, 128)
        self.up4 = nn.ConvTranspose2d(128, 64, kernel_size=2, stride=2)
        self.conv4 = DoubleConv(128, 64)
        self.outc = nn.Conv2d(64, n_classes, kernel_size=1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x1 = self.inc(x); x2 = self.down1(x1); x3 = self.down2(x2); x4 = self.down3(x3); x5 = self.down4(x4)
        y = self.up1(x5)
        diffY, diffX = x4.size()[2] - y.size()[2], x4.size()[3] - y.size()[3]
        y = F.pad(y, [diffX // 2, diffX - diffX // 2, diffY // 2, diffY - diffY // 2])
        x = self.conv1(torch.cat([x4, y], dim=1))
        y = self.up2(x)
        diffY, diffX = x3.size()[2] - y.size()[2], x3.size()[3] - y.size()[3]
        y = F.pad(y, [diffX // 2, diffX - diffX // 2, diffY // 2, diffY - diffY // 2])
        x = self.conv2(torch.cat([x3, y], dim=1))
        y = self.up3(x)
        diffY, diffX = x2.size()[2] - y.size()[2], x2.size()[3] - y.size()[3]
        y = F.pad(y, [diffX // 2, diffX - diffX // 2, diffY // 2, diffY - diffY // 2])
        x = self.conv3(torch.cat([x2, y], dim=1))
        y = self.up4(x)
        diffY, diffX = x1.size()[2] - y.size()[2], x1.size()[3] - y.size()[3]
        y = F.pad(y, [diffX // 2, diffX - diffX // 2, diffY // 2, diffY - diffY // 2])
        x = self.conv4(torch.cat([x1, y], dim=1))
        return self.sigmoid(self.outc(x))

# 2. Dataset Loader
class SatelliteWaterDataset(Dataset):
    def __init__(self, image_dir, mask_dir, img_size=(256, 256)):
        self.image_dir, self.mask_dir = image_dir, mask_dir
        self.img_size = img_size
        self.images = [f for f in os.listdir(image_dir) if f.endswith('.jpg') or f.endswith('.png')]

    def __len__(self): return len(self.images)

    def __getitem__(self, idx):
        img_name = self.images[idx]
        image = Image.open(os.path.join(self.image_dir, img_name)).convert("RGB")
        mask = Image.open(os.path.join(self.mask_dir, img_name)).convert("L")

        image = TF.resize(image, self.img_size)
        mask = TF.resize(mask, self.img_size)

        # Simple Augmentation for Colab
        if random.random() > 0.5: image, mask = TF.hflip(image), TF.hflip(mask)

        image, mask = TF.to_tensor(image), TF.to_tensor(mask)
        return image, (mask > 0.5).float()

# 3. Training Loop
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Training on: {device}")

# If you unzipped data to a specific folder, update these paths
img_dir, mask_dir = "./data_folder/images", "./data_folder/masks"

if os.path.exists(img_dir):
    model = WaterloggingUNet().to(device)
    criterion = nn.BCELoss()
    optimizer = optim.Adam(model.parameters(), lr=1e-4)

    dataset = SatelliteWaterDataset(img_dir, mask_dir)
    train_size = int(0.8 * len(dataset))
    train_data, val_data = random_split(dataset, [train_size, len(dataset) - train_size])
    train_loader = DataLoader(train_data, batch_size=16, shuffle=True)
    val_loader = DataLoader(val_data, batch_size=16)

    for epoch in range(15):
        model.train()
        epoch_loss = 0
        for img, mask in train_loader:
            img, mask = img.to(device), mask.to(device)
            optimizer.zero_grad()
            loss = criterion(model(img), mask)
            loss.backward(); optimizer.step()
            epoch_loss += loss.item()

        print(f"Epoch [{epoch+1}/15] | Loss: {epoch_loss/len(train_loader):.4f}")

    torch.save(model.state_dict(), 'waterlogging_unet.pth')
    print("Training Done! File saved as waterlogging_unet.pth")
    from google.colab import files
    files.download('waterlogging_unet.pth')
else:
    print(f"Error: Dataset not found in {img_dir}. Please ensure your zip extraction matched this folder structure.")
```
